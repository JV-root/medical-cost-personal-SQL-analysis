```python
import pandas as pd
import sqlite3

df = pd.read_csv("insurance.csv")
conn = sqlite3.connect("insurance.db")
df.to_sql("insurance", conn, if_exists="replace", index=False)

df.head()
```

```python
%load_ext sql
%sql sqlite:///insurance.db
```

```python
%%sql
--- 1- Qual a média de BMI entre homens e mulheres? existe uma diferença significativa? 
SELECT sex AS sexo, ROUND(AVG(bmi), 2) AS Bioimpedancia
FROM insurance
GROUP BY sex
```

```python
%%sql
SELECT * 
FROM insurance
```

```python
%%sql
SELECT
    CASE
        WHEN age > 30 THEN 'Acima de 30'
        ELSE '30 ou menos'
    END AS grupo_idade,
    ROUND(AVG(bmi), 2) AS Media_BMI,
    COUNT(*) AS quantidade_pessoas
FROM insurance
GROUP BY grupo_idade;
```
