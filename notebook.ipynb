```python
%%sql
SELECT
    CASE
        WHEN age > 30 THEN 'Acima de 30'
        ELSE '30 ou menos'
    END AS grupo_idade,
    sex AS sexo,
    ROUND(AVG(bmi), 2) AS Media_BMI,
    COUNT(*) AS quantidade_pessoas
FROM insurance
GROUP BY grupo_idade, sex
ORDER BY Media_BMI DESC;
```

```python
%%sql
SELECT 
    CASE
        WHEN smoker = 'yes' THEN 'Fumante'
        ELSE 'Não fumante'
    END as status_fumante,
    ROUND(AVG(bmi), 2) AS Media_BMI,
    COUNT (*) AS quantidade_pessoas
FROM insurance
GROUP BY status_fumante
ORDER BY Media_BMI DESC;
```

```python
%%sql
WITH min_count AS (
  SELECT COUNT(*) as min_fumantes FROM insurance 
  WHERE smoker = 'yes'
),
fumantes_amostra AS (
  SELECT * FROM insurance 
  WHERE smoker = 'yes'
  ORDER BY RANDOM() 
  LIMIT (SELECT min_fumantes FROM min_count)
),
nao_fumantes_amostra AS (
  SELECT * FROM insurance 
  WHERE smoker = 'no'
  ORDER BY RANDOM() 
  LIMIT (SELECT min_fumantes FROM min_count)
)
SELECT 
  CASE WHEN smoker = 'yes' THEN 'Fumante' ELSE 'Não fumante' END AS status,
  ROUND(AVG(bmi), 2) AS Media_BMI,
  COUNT(*) AS quantidade
FROM (SELECT * FROM fumantes_amostra UNION ALL SELECT * FROM nao_fumantes_amostra)
GROUP BY status;
```
